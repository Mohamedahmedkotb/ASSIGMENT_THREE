# Question 1

```python
import numpy as np

arr = np.random.randint(1, 101, size=(3, 3))

minimum = np.min(arr)
maximum = np.max(arr)

normalized = (arr - minimum) / (maximum - minimum)

print("Original array:")
print(arr)

print("Normalized array:")
print(normalized)
```

# Question 2

```python
import numpy as np
arr = np.random.randint(1, 101, size=10)
print("Array:", arr)
x = int(input("Enter X: "))
difference = np.abs(arr - x)
index = np.argmin(difference)
print("Closest value:", arr[index])
```

# Question 3 

```python
import numpy as np
arr = np.random.randint(1, 101, size=(5, 5))
print("Original array:")
print(arr)
sorted_arr = np.sort(arr, axis=0)
second_largest = sorted_arr[-2]
print("Second-largest values:")
print(second_largest)
```

# Question 4 

```python
import numpy as np
arr = np.random.randint(1, 101, size=(6, 4))
print("Original array:")
print(arr)
row_sum = np.sum(arr, axis=1)
indices = np.where(row_sum % 2 == 0)[0]
print("Sum of each row:", row_sum)
print("Indices of even rows:", indices)
```

# Question 5

```python
import numpy as np
arr = np.random.randint(1, 51, size=30)
print("Original array:")
print(arr)
arr[(arr % 3 == 0) | (arr % 5 == 0)] = -1
print("Modified array:")
print(arr)
```

# Question 6 

```python
import numpy as np
arr = np.random.randint(1, 101, size=(3, 3))
print("Original array:")
print(arr)
row_mean = np.mean(arr, axis=1, keepdims=True)
result = arr - row_mean
print("Mean of each row:")
print(row_mean)
print("Result:")
print(result)
```
